```python
taku = wgms.loc[wgms.NAME == 'TAKU']
for site_name, df in taku.groupby('POINT_ID'):
    if len(df) < 10:
        continue
    fig, ax = plt.subplots(1,1)
    ax.plot(df.YEAR, df.POINT_BALANCE * df.DENSITY, c='red', marker='o')
    ax.set_title(site_name)
    
```
